```python
import zmq

ok = 0
fail = 0
errors = []

ctx = zmq.Context()

for i in range(1000):
    s = ctx.socket(zmq.ROUTER)
    try:
        s.bind("tcp://127.0.0.1:*")
        ok += 1
    except Exception as e:
        fail += 1
        errors.append(str(e))
    finally:
        s.close(linger=0)

ctx.term()

print("OK =", ok)
print("FAIL =", fail)

if errors:
    print("FIRST ERROR =", errors[0])
```
